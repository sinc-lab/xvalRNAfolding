```python
import pandas as pd
from sklearn.model_selection import KFold, train_test_split

data_path = "../../../data/bpRNA/"
df = pd.read_csv(data_path + "bpRNA.csv", index_col=0)
display(df.head(1))
print(f"Number of sequences: {len(df)}")
```

```python
k_folds = 5
kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)

splits = []
for fold_number, (train_val_idx, test_idx) in enumerate( kf.split(df) ):

    train_val = df.iloc[train_val_idx]
    test = df.iloc[test_idx]

    train, valid = train_test_split(train_val,test_size=0.10, random_state=42)

    fold_split = pd.DataFrame({"fold_number": fold_number, "partition": "train", "id": df.index})

    fold_split.loc[fold_split["id"].isin(valid.index),"partition"] = "valid"
    fold_split.loc[fold_split["id"].isin(test.index),"partition"] = "test"

    splits.append(fold_split)

df_splits = pd.concat(splits, ignore_index=True)
df_splits
```

```python
df_splits.groupby(["fold_number", "partition"]).count() 
```

```python
df_splits[["id", "fold_number", "partition"]].to_csv(data_path + "random_kfold_bprna.csv", index=False)
```
