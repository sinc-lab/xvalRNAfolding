Extract the partitions form directories to CSV format

Dataset: ArchiveII
  https://rna.urmc.rochester.edu/pub/archiveII.tar.gz

Splits from:
  https://github.com/marcellszi/dl-rna/releases

Paper:
  https://doi.org/10.1093/bioinformatics/btac415
  https://github.com/marcellszi/dl-rna

```python
import os
import glob
import pandas as pd
from tqdm import tqdm
```

```python
DATA_DIR = "/DATA/dl-rna/ct-splits/ct/"
```

```python
arch2 = glob.glob(DATA_DIR + "**/*/" + "*.ct", recursive=True)

fold_numbers = {} # to convert fold names in fold numbers
n_fam_folds = 0   # folds converted to number in fam-fold splits
all_splits = pd.DataFrame(columns=["split_type", "fold_name", "fold_number", "partition", "id"])
for i, fname in enumerate(tqdm(arch2)):
    pathn = os.path.normpath(fname[len(DATA_DIR):])
    pathx = pathn.split(os.sep)
    
    fold_name = pathx[1]
    if pathx[0] == "fam-fold":
        if fold_name in fold_numbers:
            fold_number = fold_numbers[fold_name]
        else:
            fold_numbers[fold_name] = n_fam_folds
            fold_number = n_fam_folds
            n_fam_folds += 1
    else:
        fold_number = int(pathx[1])
        
    all_splits.loc[i] = [pathx[0], fold_name, fold_number, pathx[2], pathx[3][:-3]]

all_splits
```

```python
all_splits.to_csv("./data/archiveII_splits.csv", index=False)
```
