```python
#  train 90 / validation 10
# ------------- reference head -------------
# fold_number,partition,id
# 0,train,tmRNA_Stre.gord._TRW-29390_1-349

import pandas as pd
from sklearn.model_selection import StratifiedKFold, train_test_split
df = pd.read_csv("archiveII_250808.csv", index_col="id")
df["fam"] = df.index.str.split("_").str[0]
df.head(1)
```

```python
k_folds = 5
skf = StratifiedKFold(n_splits=k_folds, shuffle=True, random_state=42)

splits = []
for fold_number, (train_val_idx, test_idx) in enumerate( skf.split(df, df["fam"]) ):

    train_val = df.iloc[train_val_idx]
    test = df.iloc[test_idx]

    train, valid = train_test_split(train_val,test_size=0.10,stratify=train_val["fam"],random_state=42)

    fold_split = pd.DataFrame({"fold_number": fold_number, "partition": "train", "id": df.index})

    fold_split.loc[fold_split["id"].isin(valid.index),"partition"] = "valid"
    fold_split.loc[fold_split["id"].isin(test.index),"partition"] = "test"

    splits.append(fold_split)

df_splits = pd.concat(splits, ignore_index=True)
```

```python
df_splits.to_csv("stratified_kfold_archiveII_250808.csv", index=False)
```
